# 1. dataclasses: Less boilerplate, more clarity


````python
from dataclasses import dataclass


@dataclass
class Server:
    host: str
    port: int = 8000
    use_ssl: bool = False

server = Server(host="api.example.com")
print(server)
````

````python
from dataclasses import dataclass


@dataclass(frozen=True)
class Point:
    x: float
    y: float
````

# 3. functools: Caching, partial functions, and other useful machinery


````python
from functools import cache

@cache
def power(base: int, exponent: int) -> int:
    print(f"Computing {base}^{exponent}")
    return base**exponent

print(power(2, 10))
print(power(3, 5))
print(power(2, 10))
````

````python
from functools import partial

def power(base: int, exponent: int) -> int:
    return base**exponent

square = partial(power, exponent=2)
cube = partial(power, exponent=3)
print(square(5))
print(cube(5))
````

````python
def square(base: int) -> int:
    return power(base, 2)
````

# 5. graphlib: Put dependency-based tasks in the correct order


````python
from graphlib import TopologicalSorter

dependencies = {
    "fetch_sources": set(),
    "compile": {"fetch_sources"},
    "test": {"compile"},
    "package": {"test"},
    "deploy": {"package"},
}
sorter = TopologicalSorter(dependencies)
for task in sorter.static_order():
    print(task)
````

````python
from graphlib import TopologicalSorter

dependencies = {
    "fetch_sources": set(),
    "compile_backend": {"fetch_sources"},
    "build_frontend": {"fetch_sources"},
    "run_tests": {"compile_backend", "build_frontend"},
    "package": {"run_tests"},
    "deploy": {"package"},
}
order = list(TopologicalSorter(dependencies).static_order())
print(order)
````

# 6. heapq: A priority queue without repeatedly sorting everything



````python
import heapq

tasks = [
    (3, "Send email to client"),
    (2, "Write documentation"),
    (1, "Fix critical bug"),
]
heapq.heapify(tasks)
while tasks:
    priority, task = heapq.heappop(tasks)
    print(priority, task)
````

````python
import heapq
tasks = [
    (3, "Send email to client"),
    (2, "Write documentation"),
    (1, "Fix critical bug"),
]
heapq.heapify(tasks)
while tasks:
    priority, task = heapq.heappop(tasks)
    print(f"Processing: {task}")
    if task == "Fix critical bug":
        print("New urgent task arrived")
        heapq.heappush(tasks, (0, "Deploy hotfix"))
    if task == "Write documentation":
        heapq.heappush(tasks, (4, "Refactor old module"))
````

# 9. textwrap: Make long text readable


````python
import textwrap

message = (
    "Python's standard library contains many modules that can "
    "remove dependencies and simplify small automation scripts."
)
print(textwrap.fill(message, width=40))
````

````python
indented = textwrap.indent(
    message,
    prefix="> ",
)

print(indented)
````

````python
summary = textwrap.shorten(
    message,
    width=50,
    placeholder="...",
)
print(summary)
````

# 10. itertools: Efficient tools for iteration


````python
from itertools import combinations

items = ["A", "B", "C"]
pairs = list(combinations(items, 2))
print(pairs)
````

````python
from itertools import count

counter = count(start=10, step=5)
print(next(counter))
print(next(counter))
print(next(counter))
````

````python
from itertools import cycle

states = cycle(["ON", "OFF"])
for _ in range(6):
    print(next(states))
````

````python
from itertools import groupby

values = ["A", "A", "B", "B", "B", "C", "A"]
groups = [
    (key, list(group))
    for key, group in groupby(values)
]
print(groups)
````
